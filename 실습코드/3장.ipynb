```python
import pandas as pd
import numpy as np
import os

# 1. 행 수 정의 (n = 100,000)
n = 100000

# 2. time: 2026-03-01부터 1분 간격으로 10만 개 생성
time_range = pd.date_range(start='2026-03-01', periods=n, freq='1min')

# 무작위 수 생성을 위한 시드 고정 (재현 가능하도록)
np.random.seed(42)

# 3. 각 컬럼별 정규분포 데이터 생성 및 반올림
# temp: 평균 450, 표준편차 5
temp_values = np.random.normal(loc=450, scale=5, size=n)

# pressure: 평균 1.0, 표준편차 0.05
pressure_values = np.random.normal(loc=1.0, scale=0.05, size=n)

# vibration: 평균 0.3, 표준편차 0.1
vibration_values = np.random.normal(loc=0.3, scale=0.1, size=n)

# 4. 데이터프레임 구성
df_equipment = pd.DataFrame({
    'time': time_range,
    'temp': np.round(temp_values, 2),
    'pressure': np.round(pressure_values, 3),    # 압력은 소수점 셋째 자리까지
    'vibration': np.round(vibration_values, 3)   # 진동도 소수점 셋째 자리까지
})

# 5. 저장 폴더 생성 및 CSV 파일 저장
os.makedirs('data', exist_ok=True)
df_equipment.to_csv('data/huge.csv', index=False)

# 결과 확인
print("--- 생성된 데이터 앞부분 (head) ---")
print(df_equipment.head())
print(f"\n성공적으로 'data/huge.csv'에 {len(df_equipment)}행 데이터가 저장되었습니다!")
```

```python
# 260615huge.py
# 대용량 CSV 생성 + 나눠 읽기 실습
from pathlib import Path
import pandas as pd
import numpy as np

csv_path = "data/huge.csv"

# 4. chunksize로 나눠 읽기
total_temp = 0
row_count = 0
chunk_count = 0

for chunk in pd.read_csv(csv_path, chunksize=10000):
    chunk_count += 1
    row_count += len(chunk)
    total_temp += chunk["temp"].sum()

    print(f"{chunk_count}번째 chunk 처리 완료 / 행 개수: {len(chunk)}")

print("전체 chunk 개수:", chunk_count)
print("전체 행 개수:", row_count)
print("전체 온도 합:", total_temp)
print("전체 온도 평균:", total_temp / row_count)

print("\n" + "=" * 50)
print("2. dtype을 지정해서 CSV 읽기")
print("=" * 50)

# 5. 자료형을 미리 지정해서 메모리 절약
df_small = pd.read_csv(
    csv_path,
    dtype={
        "temp": "float32",
        "pressure": "float32",
        "vibration": "float32"
    }
)

print(df_small.info())
print(df_small.head())
print("전체 행 개수:", len(df_small))
print("온도 평균:", df_small["temp"].mean())
print("압력 평균:", df_small["pressure"].mean())
print("진동 평균:", df_small["vibration"].mean())

print("\n실습 완료")
```

```python
import glob, pandas as pd

files = glob.glob("data/log_2026-03-*.csv")

df_list = []
for f in files:
    temp = pd.read_csv(f)
    temp['source'] = f
    df_list.append(temp)

merged = pd.concat(df_list, ignore_index = True)
print(merged.shape)

print("\n앞 5행 확인")
print(merged.head())

print("\nsource 컬럼만 확인")
print(merged[['time', 'temp', 'source']].head(10))

print("\n어떤 파일들이 합쳐졌는지 확인")
print(merged['source'].unique())

print("\n파일별 행 개수 확인")
print(merged['source'].value_counts().sort_index())


```
